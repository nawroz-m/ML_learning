# Ensemble Methods

## 1. Boosting

**Goal:** Convert **weak learners → strong learner**.

Key idea:

* Models are trained **sequentially**.
* Each new model focuses **more on previous mistakes**.
* Final prediction = **weighted combination of all models**.

---

# 2. AdaBoost

**Idea:** Focus learning on **misclassified samples**.

Steps:

1. Train a weak classifier.
2. Increase weight of **misclassified samples**.
3. Train next classifier using updated weights.
4. Repeat.
5. Combine all classifiers with **weighted voting**.

Classifier weight:

$$
\alpha_t = \frac{1}{2}\ln\left(\frac{1-\epsilon_t}{\epsilon_t}\right)
$$

Where:

* **αₜ** = weight of classifier
* **εₜ** = error rate

Lower error → **higher classifier weight**.

Final model = **weighted sum of weak classifiers**.

---

# 3. Random Forest

Ensemble of **decision trees**.

Key ideas:

* Train each tree on **random samples (bagging)**.
* At each split use **random subset of features**.
* Final prediction:

  * **Majority vote** (classification)
  * **Average** (regression)

Purpose: **reduce overfitting and variance**.

---

# 4. Stacking

Combine **different models** using a **meta-model**.

Process:

1. Train base models (e.g., KNN, SVM, Tree).
2. Use their **predictions as features**.
3. Train a **meta-classifier** to make final prediction.

Important:

* Meta-model must use **predictions from unseen data** (e.g., cross-validation) to avoid **overfitting**.

---

✅ **One-line summary**

* **Boosting** → sequential error correction
* **AdaBoost** → weighted samples
* **Random Forest** → random trees
* **Stacking** → model on top of models
