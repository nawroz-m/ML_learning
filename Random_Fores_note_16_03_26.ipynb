## **Random Forest (RF) Notes**

**Concept:**
Random Forest is an ensemble learning method that builds multiple decision trees (N trees) for classification or regression and combines their outputs to improve accuracy and reduce overfitting.

**Key Components:**

1. **Estimator (Number of Trees)**

   * Each tree is called an *estimator*.
   * `n_estimators` is the number of trees in the forest. More trees → usually better performance but slower training.
   * Goal: find the optimal number of trees for your dataset.

2. **Splitting Criteria**

   * Common options: **Gini impurity** or **Entropy**.
   * Determines how the tree splits at each node.

3. **Feature Selection (Randomness in RF)**

   * At each split, RF selects the *best feature* from a random subset of features.
   * Helps reduce correlation between trees → improves generalization.
   * You can control this with parameters like `max_features`.

4. **Data Sampling (Bootstrap)**

   * RF uses *bootstrap sampling*: each tree is trained on a random subset of data (with replacement).
   * *Shuffling* ensures randomness.
   * This is similar to a form of cross-validation within the forest (Out-of-Bag error can be used to estimate accuracy).

**Advantages vs. Single Decision Tree:**

* Reduces overfitting compared to a single tree.
* Handles high-dimensional data well.
* Provides feature importance rankings automatically.
* More robust to noisy data.

**Training Task:**

* Use RF instead of SVM.
* Steps:

  1. Read the dataset carefully.
  2. Decide parameters: `n_estimators`, `max_features`, `criterion`, etc.
  3. Use grid search or random search to find optimal configuration.
  4. Train the model on the dataset.
  5. Compare results with SVM for evaluation.

**Notes on Optimization:**

* Parameter tuning is key: number of trees, depth, features per split.
* Shuffle your data before training to avoid bias.
* Use cross-validation (or Out-of-Bag error) to evaluate model performance.

**Final Comparison:**

* Train two models: **SVM vs Random Forest**.
* Compare metrics like accuracy, F1-score, or ROC-AUC to decide which model works better for your task.
