## 📌 Data Characteristics

---

  ### 🔹 1. Dimensionality

  * Number of features (columns).
  * High dimension → high complexity.
  * Problems:

    * Overfitting
    * High computation
    * Curse of dimensionality

  👉 Solution: Feature selection / dimensionality reduction.

  ---

  ### 🔹 2. Sparsity

  * Most values = 0.
  * Common in text & recommender systems.
  * Store only non-zero values.

  👉 Saves memory & speed.

  ---

  ### 🔹 3. Resolution

  * Level of detail in data.
  * High resolution → detailed.
  * Low resolution → averaged/summarized.
  * Patterns depend on scale.

  ---

  ### 🔹 4. Size

  * Number of data objects (rows).
  * Large data → scalable algorithms needed.
  * Small data → overfitting risk.

  ---

  ## 📂 Types of Data

  #### 1️⃣ Record Data

  * Data matrix (rows × columns)
  * Document data → term vectors
  * Transaction data → set of items per transaction

  #### 2️⃣ Graph Data

  * Nodes + edges
  * Relationships matter

  #### 3️⃣ Ordered Data

  * Order is important
  * Time-series
  * Sequential data
  * Spatio-temporal data

  ---

  ## 📊 Data Quality

  ### 🔹 Missing Data

  * Very common (e.g., healthcare)
  * Solutions:

    * Remove
    * Impute (mean/median/model)

  ### 🔹 Noise

  * Random errors
  * Object noise → wrong sample
  * Attribute noise → wrong feature value

  ### 🔹 Outliers

  * Very different from most data
  * Can be:

    * Error
    * Important rare event

    Applications:

    * Fraud detection
    * Intrusion detection
    * Medical anomalies
