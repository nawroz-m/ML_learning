## 1️⃣ Gradient Descent

### 🔹 What is Gradient Descent?

**Gradient Descent (GD)** is an optimization algorithm used to find the **minimum value of a loss function**.

👉 Goal:
Find the best model parameters (weights) that **minimize the loss**.

---

### 🔹 Loss Function

A **loss function** measures how well the model predicts the target values.

It tells us:

> "How wrong is the model?"

#### Common Loss Functions

1. **Mean Squared Error (MSE)**
   $$
   MSE = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2
   $$

Where:

* $ N $ = number of samples
* $ y_i $ = true value
* $ \hat{y}_i $ = predicted value

✔ Used in regression
✔ Penalizes large errors more (because of square)

---

2. **Mean Absolute Error (MAE)**
   $$
   MAE = \frac{1}{N} \sum_{i=1}^{N} |y_i - \hat{y}_i|
   $$

✔ Also used in regression
✔ Less sensitive to large outliers than MSE

---

#### 🔹 Parametric Function

A **parametric function** is a function that depends on parameters (weights).

Example:
$$
y = w x + b
$$

Where:

* $ w $ = weight
* $ b $ = bias

❗ It is NOT the same as the optimization function.

* Parametric function → defines the model
* Optimization function → updates the parameters to minimize loss

---

### 2️⃣ Forward & Backward Propagation

#### 🔹 Forward Propagation

Steps:

1. Input data enters the model
2. Model computes prediction
3. Loss is calculated

👉 This is the "prediction phase".

---

#### 🔹 Backpropagation

Backpropagation computes the **derivative (gradient)** of the loss with respect to the weights.

$$
\frac{\partial Loss}{\partial w}
$$

It tells us:

> "How should we change the weights to reduce loss?"

The gradient is propagated backward through the network.

---

### 3️⃣ Training Concepts

#### 🔹 1. Weights

Weights are the parameters learned during training.

They are updated using:

$$
w = w - \eta \frac{\partial Loss}{\partial w}
$$

Where:

* $ w $ = weight
* $ \eta $ = learning rate
* $ \frac{\partial Loss}{\partial w} $ = gradient

---

### 🔹 2. Convergence

Convergence happens when:

* Loss stops decreasing
* Updates become very small

---

### 🔹 3. Learning Rate (η)

Learning rate controls how big the update step is.

* Too large → training unstable
* Too small → very slow learning

👉 Always monitor the loss value during training.

---

### 🔹 4. Overfitting

Overfitting = model learns training data too well (including noise).

Result:

* Low training loss
* High validation loss

#### Solutions:

* Regularization (L1, L2)
* Simpler model
* More data
* Early stopping

---

### 🔹 5. Underfitting

Underfitting = model is too simple.

Result:

* High training loss
* High validation loss

👉 Always use a **validation set** to detect overfitting or underfitting.

---

### 🔹 6. Activation Functions

Activation functions introduce **non-linearity**.

Without them → model is only linear.

#### Common Activation Functions:

#### 1. Sigmoid

$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$

Output range: (0,1)

Used in:

* Binary classification

---

#### 2. ReLU (Rectified Linear Unit)

$$
ReLU(x) = \max(0, x)
$$

Output range: [0, ∞)

✔ Very popular
✔ Computationally efficient
✔ Helps with deep networks

---

# 4️⃣ Training Procedure

#### Step 1

Initialize weights (usually small random values)

#### Step 2

Repeat until:

* Convergence OR
* Maximum epochs reached

Inside loop:

1. Forward propagation
2. Compute loss
3. Backpropagation
4. Update weights

Often using:

**Stochastic Gradient Descent (SGD)**
→ Updates weights using small mini-batches.

---

# 📘 Data Preprocessing




Data preprocessing = transforming raw data into a clean, numerical, and well-scaled format suitable for ML training.

---

## 1️⃣ Data Cleaning

### 🔹 Missing Data Handling

Let $ X $ be the dataset.

**Options:**

1. **Drop feature (column)**

   * If high percentage of missing values
   * If low predictive importance

2. **Drop sample (row)**

   * If few missing values
   * Risk: reduces dataset size

3. **Imputation (Most common)**
   Replace missing value $ x_i $ with:

   * Mean → numerical features
   * Median → robust to outliers
   * Mode → categorical features
   * Constant
   * Model-based estimate

Goal: preserve dataset size while minimizing bias.

---

## 2️⃣ Encoding (Categorical → Numerical)

ML models require numerical input.

### 🔹 Label Encoding

$$
\text{Category} \rightarrow {0,1,2,...}
$$

* Efficient
* Introduces artificial ordering

Used for ordinal categories.

---

### 🔹 One-Hot Encoding

Transforms categorical variable with $ k $ categories into $ k $ binary features.

* No artificial order
* Increases dimensionality

Preferred for nominal categories.

---

## 3️⃣ Feature Scaling

Features with different ranges distort distance-based or gradient-based algorithms.

---

### 🔹 Min-Max Normalization

$$
x' = \frac{x - x_{min}}{x_{max} - x_{min}}
$$

Maps feature to range ([0,1]).

* Preserves distribution shape
* Sensitive to outliers

Used when range is known.

---

### 🔹 Standardization (Z-score)

$$
x' = \frac{x - \mu}{\sigma}
$$

Where:

* $ \mu $ = mean
* $ \sigma $ = standard deviation

Result:
$$
\mu = 0,\quad \sigma^2 = 1
$$

* Not bounded
* Robust for many ML models
* Required for SVM, Logistic Regression, Neural Networks

---

## 4️⃣ Dataset Splitting

Split dataset into:

* **Training set** → parameter learning
* **Validation set** → hyperparameter tuning
* **Test set** → final evaluation

Typical ratio:
$$
70% / 15% / 15%
$$

Test set must remain unseen.

---

## 5️⃣ Outlier Detection

Outliers = samples with low probability under data distribution.

---

### 🔹 Elliptic Envelope

* Assumes Gaussian distribution
* Fits covariance-based ellipse
* Points outside → outliers

Parametric method.

---

### 🔹 Kernel Density Estimation (KDE)

Estimates probability density:

$$
\hat{f}(x) = \frac{1}{n} \sum_{i=1}^{n} K(x - x_i)
$$

Low-density points → outliers.

Kernel options:

* Histogram (discrete approximation)
* Gaussian kernel (smooth density)

Non-parametric method.

---

### 🔹 Isolation Forest

* Random tree partitioning
* Outliers require fewer splits to isolate
* No distribution assumption
* Works well in high dimensions

---

# 🌳 Decision Tree

Supervised learning model.

Splits data by minimizing impurity.

Common criteria:

* **Gini Index**
* **Entropy**

High variance model → prone to overfitting.

---

# 🌲 Random Forest

Ensemble of Decision Trees.

Algorithm:

1. Bootstrap sampling
2. Train tree on each sample
3. Random feature selection at each split
4. Aggregate predictions:

   * Classification → majority vote
   * Regression → mean

Advantages:

* Reduces variance
* Improves generalization
* Less overfitting than single tree

---
