### Learning Types

**Supervised Learning**

* Predict output using **labeled data**.
* Types:

  * **Classification** → categorical output
  * **Regression** → numerical output

**Unsupervised Learning**

* No labels.
* Goal: **find structure in data**.
* Example: **Clustering**

---

## Goal of Machine Learning

**Generalization**

A good model should:

* Work well on **unseen data**
* Have **low prediction error**

---

## Dataset Split

**Training Set**

* Train the model.

**Validation Set**

* Tune **hyperparameters**.

**Test Set**

* Final **performance evaluation**.

Important:
All sets must be **separate (non-overlapping)**.

---

## Prediction Error

Compare **prediction vs ground truth**.

Indicator function:

$$
I(y \neq \hat{y})
##

* **y** = true label
* **ŷ** = predicted label

Result:

* **0 → correct**
* **1 → error**

Goal → **minimize average error**

---

## Main Learning Algorithms

#### Linear Classifier

* Uses **weights (w)** and **bias (b)**.
* Creates a **linear decision boundary** (line/plane).

Goal:

* Separate classes.

---

#### Decision Tree

Structure:

* **Root**
* **Nodes**
* **Leaves**

Popular algorithms:

* **ID3 → Entropy**
* **C4.5 → Gain Ratio**
* **CART → Gini Index**

Problem:

* **Overfitting**

Solution:

* **Pruning**

  * Pre-pruning
  * Post-pruning

Boundary:

* **Axis-aligned (x or y direction)**

---

#### Neural Networks (NN)

Neuron contains:

* **Weights**
* **Bias**
* **Activation function**

Activation adds **non-linearity**.

Training method:
**Backpropagation**

* compute error
* update weights using gradients

---

#### K-Nearest Neighbors (KNN)

Idea:

* Similar inputs → similar labels.

Characteristics:

* **Lazy learning**
* Stores training data.

Hyperparameters:

* **K (number of neighbors)**
* **Distance metric**
* **Voting method**

---

#### Support Vector Machine (SVM)

Goal:

* Find **hyperplane with maximum margin**.

Key concept:

* **Support vectors** (closest points)

Loss:

* **Hinge Loss**

For non-linear problems:

* Use **Kernel functions**
* Example: **Gaussian (RBF) Kernel**

---

## Model Evaluation

#### Validation Set

Used for **hyperparameter tuning**.

---

#### Cross-Validation

**K-Fold Cross Validation**

Steps:

1. Split data into **K parts**
2. Train on **K−1**
3. Test on **1**
4. Repeat **K times**

Final result = **average score**
