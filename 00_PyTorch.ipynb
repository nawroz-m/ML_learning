```python
# Shuffle tensore using merge indexes
t=torch.tensor([[1,2],[3,4]])
r=torch.randperm(2)
c=torch.randperm(2)
t=t[r[:, None], c]
t
```

```python
t.nelement()
```

```python
# Suffle tensor using view the best practice
indx = torch.randperm(t.nelement())
t.view(-1)[indx].view(t.size())
```
