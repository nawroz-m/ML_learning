### 1. Normalize all signals to the range [0, 1] and plot them.

```python
# 1. Get all the normalized dataset between 0 & 1, we will ingnore the NaN
norm_btw_0_1 = z_normalized_data[(z_normalized_data>=0) & (z_normalized_data<=1)][:4]
norm_btw_0_1.head(3)
```

```python
#2. Plot normalized signal range between [0, 1]
plt.figure(figsize=(14, 6))
plt.plot(norm_btw_0_1)
plt.grid()
plt.title("Normalized signals between [0, 1]")
plt.xlabel("Object numbers")
plt.ylabel("Amplitued")
```

### 2. Compute the mean and standard deviation of the normalized data.

```python
# Compute the Mean of normalized data
mean = norm_btw_0_1.mean()
mean[4:5]
```

```python
# Compute the Standard deviation of normalized data
std = norm_btw_0_1.std()
std[12:14]
```

### 3. Plot the mean & standard deviation of the normalized dataset.

```python
mean.head(2)
```

```python
mean.shape
```

```python
# 1. Plot the mean
plt.figure(figsize=(15, 6))
plt.plot(mean.values, 'b-', label="Mean")
plt.grid()
plt.title("Mean of the normalized")
plt.xlabel("Number of objects")
plt.ylabel("Amplitued")
plt.legend()
```

```python
# 2. Plot the standard deviation of the normalized dataset
plt.figure(figsize=(15, 6))
plt.plot(std.values, 'y-', label="Standard deviation")
plt.grid()
plt.title("Standard deviation of normalized dataset between [0, 1]")
plt.xlabel("Object no")
plt.ylabel("Amplitude")
plt.legend()
```

```python
# Plot both together
plt.figure(figsize=(15, 6))
plt.plot(mean.values, "b-", label="Mean",)
plt.plot(std.values, "y-", label="Standard Deviation") # alpha=opacity
plt.title("Mean & Standard Deviation of normalized data", alpha=0.3)
plt.xlabel("Object number")
plt.ylabel("Amplitude")
plt.grid()
plt.legend()
```

### 4. **Identify potential outlier signals**, defined as signals that lie outside the mean ± standard deviation, and plot them
https://kmr-gautam2893.medium.com/how-to-identify-the-outliers-in-your-data-ee9c28b42fc3

image.png

```python
# 1. Get all the outlier signals
outliers = norm_btw_0_1[(norm_btw_0_1<(mean-std)) | (norm_btw_0_1>(mean+std))]
outliers.head(2)
```

```python
# 2. Plot the outlier signals
plt.figure(figsize=(15, 6))
plt.plot(outliers)
plt.title("Outliers")
plt.xlabel("Object number")
plt.ylabel("Amplitude")
plt.grid()
```

### 5. **Compute the distance of each signal from the mean in terms of standard deviation.**

image.png

**Meaning**

* Z = 0 → value is exactly at the mean

* Z = 1 → value is 1 standard deviation above the mean

* Z = -1 → value is 1 standard deviation below the mean

**Why this is used**

This tells you how unusual a value is.

Example:

* |Z| ≤ 1 → normal value

* |Z| > 1 → possible outlier

```python
# 5. Compute the distance of each signal from the mean in terms of standard deviation.
z_score = (norm_btw_0_1 - mean)/std
```

```python
plt.figure(figsize=(15, 8))
plt.plot(z_score)
plt.title("Distance from mean")
plt.xlabel("Object no")
plt.ylabel("Amplitude")
```

### 6. **Classify a signal as an outlier if its distance is greater than a chosen threshold.** The threshold can be set to 1 or any other value depending on the analysis.

```python
z_score[:3]
```

```python
# 1. Get the outlier boolean
threshold = 0.7
otl_bool = abs(z_score)>threshold
```

```python
otl_bool.describe()
```

```python
# 2. Plot the outliere boolean to get to know more
plt.figure(figsize=(12, 6))
plt.plot(otl_bool)
plt.title("Outlier boolean")
```

```python
# 3. Get the actual outliers
outliers_final = norm_btw_0_1[otl_bool]
outliers_final.sum().sum()
```

```python
outliers_final[:4]
```

```python
# 4. Plot the actual outliers
plt.figure(figsize=(15, 6))
plt.plot(outliers_final)
plt.title("Outlier of threshold 0.7")
plt.xlabel("Object no")
plt.ylabel("Amplitude")
```
