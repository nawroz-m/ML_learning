#Image Processing Basics

## 1. What is an Image?

An **image** is a **2D grid (matrix) of numbers**.

* Each element is a **pixel**.
* Each pixel stores an **intensity value** (brightness or color).

Example (grayscale image):

| Pixel  | Intensity   |
| ------ | ----------- |
| Dark   | small value |
| Bright | large value |

Typical grayscale range:

* **0 → black**
* **255 → white**

---

## 2. Digitization of Images

Images become digital through **digitization**, which includes:

### Sampling

* Divides the image into **pixels**.
* The **pixel size** is the smallest unit of information (**quantum**).

### Quantization

* Converts continuous intensity values into **discrete levels**.

Example:

* **8-bit image → 256 levels (0–255)**

Reducing bit depth means:

* fewer intensity levels
* **loss of information**

---

## 3. Image Acquisition

Images are captured by a **sensor**.

Process:

1. **Light hits the sensor**
2. Sensor converts light to **electrical energy**
3. Values are stored as **pixel intensities**

Image properties:

* **Width × Height** → spatial size
* **Depth (bit depth)** → number of intensity levels

Example:

* 8-bit → 256 intensity values.

---

## 4. Histogram

A **histogram** shows the **distribution of pixel intensities** in an image.

* X-axis → intensity values (0–255)
* Y-axis → number of pixels with that intensity

Uses:

* analyze brightness
* analyze contrast
* detect exposure problems

Example interpretation:

* histogram concentrated left → **dark image**
* histogram concentrated right → **bright image**

---

## 5. Exposure

**Exposure** describes how much light reaches the sensor.

Using histogram:

* left → **underexposed**
* right → **overexposed**
* spread across range → **well exposed**

---

## 6. Contrast

**Contrast** is the difference between **dark and bright pixels**.

Types:

* **Low contrast** → image looks flat
* **High contrast** → strong differences

Increasing contrast spreads intensity values in the histogram.

---

## 7. Dynamic Range

Dynamic range = **range between minimum and maximum intensity values used**.

Good dynamic range means:

* the image uses **most of the available intensity levels**

Reducing intensity depth:

* reduces details
* edges become less visible.

---

# Image Processing

Image processing applies **algorithms to images**.

Typical pipeline:

**Input image → algorithm → output image**

Goals:

* improve image for **human visualization**
* prepare image for **machine learning analysis**

---

# Edge Detection

An **edge** is a place where the **intensity changes suddenly**.

Example:

* border between object and background.

Edge detection helps identify:

* object boundaries
* shapes
* important structures in the image.

---

# Segmentation

**Segmentation** divides an image into **regions or objects**.

Two common approaches:

1. **Pixel-level classification**
2. **Region-based segmentation**

Goal:

* separate objects from background.

---

# Histogram Operations

## Histogram Equalization

Improves contrast by **redistributing intensity values**.

Idea:

* make histogram **more uniform**

Result:

* hidden details become more visible.

---

## Cumulative Distribution

The **cumulative histogram** shows how intensity values accumulate.

It is often used in **histogram equalization algorithms**.

---

# Point Transformations

A **point transformation** modifies **each pixel independently**.

Output pixel depends **only on the input pixel**, not neighbors.

Example operations:

* thresholding
* negative image
* intensity scaling

Useful for:

* simple segmentation
* brightness adjustment.

---

## Negative Image

Produces the **inverse intensity**.

Dark areas become bright and vice versa.

Useful for highlighting details.

---

## Gamma (Power-Law) Transformation

Adjusts brightness using a **gamma parameter**.

* **γ < 1** → enhances dark regions
* **γ > 1** → darkens the image

Used in image correction and display systems.

---

# Filters

A **filter** modifies a pixel using its **neighboring pixels**.

This uses a **kernel (small matrix)**.

Parameters:

* kernel size
* kernel shape

Filters can be:

* **linear**
* **non-linear**

---

# Convolution

Convolution applies a **filter kernel** to the image.

Process:

1. Place the kernel on a pixel
2. Multiply kernel values with neighboring pixels
3. Sum results
4. Assign value to output pixel

Mathematically it is a **double sum operation**.

---

## Handling Image Boundaries

Problems occur near edges.

Common solutions:

* **Cropping** → output becomes smaller
* **Padding** → add extra pixels
* **Extension** → repeat edge values
* **Wrap** → image repeats from opposite side

---

# Smoothing Filters

Smoothing filters **reduce noise** and small variations.

They make the image **less sharp but cleaner**.

---

## Gaussian Filter

A common smoothing filter based on a **Gaussian function**.

Parameters:

* **σ (sigma)** → controls smoothing strength
* **r** → distance from filter center

Properties:

* central pixels have **higher weight**
* far pixels have **lower weight**

Result:

* smooth image while preserving general structure.

---

# Edge Detection with Derivatives

Edges correspond to **large intensity gradients**.

Derivative filters detect these gradients.

Types:

* horizontal gradient
* vertical gradient
* gradient magnitude

Gradient magnitude highlights **strong edges**.

---

✅ **main categories:**

1. **Point operations** → work on single pixel
2. **Filters (neighborhood operations)** → use surrounding pixels
3. **Histogram methods** → modify intensity distribution
4. **Edge detection** → detect intensity changes
