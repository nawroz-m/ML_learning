● Generate 1k events gaussian distributed

● Write functions to estimate the mean and the σ of the dataset

```python
def mean(x):
  s=0
  for i in x:
    s += i
  return s/len(x)
```

```python
def standard_deviation(x):
  m = mean(x)
  varience = 0
  for i in x:
    varience += (i-m)**2
  return np.sqrt(varience/len(x))
```

```python
event_size = 1000
events = np.random.normal(loc=0, scale=1, size=event_size)
m = mean(events)
sigma = standard_deviation(events)
m, sigma
```
