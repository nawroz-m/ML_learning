# Environment – Convolutional Neural Networks (CNN)

### Digital Image

* A **digital image** is composed of **pixels**.
* Each **pixel** has:

  * **coordinates** (position in the image)
  * **intensity value** (usually **0–255**).
* A **color image** has **3 channels**:

  * **R** = Red
  * **G** = Green
  * **B** = Blue
* Image size is defined by **width × height**.

---

# Convolutional Neural Networks (CNN)

* CNNs are a **family of Machine Learning models designed to analyze images**.
* Unlike traditional neural networks, CNNs **preserve spatial relationships between pixels**.

### Why not use a normal Neural Network?

In a **Feedforward Neural Network**:

* The image must be **flattened into a vector**.
* This creates **too many parameters**.
* The model **does not scale well** and becomes **hard to train**.

CNNs solve this by **processing the image locally using filters**.

---

# Convolution (Filters)

* A **filter (kernel)** is a small matrix that **slides over the image**.
* The filter detects **patterns such as edges, textures, or shapes**.

Key ideas:

* Filters are **trainable** → the network **learns the filter weights during training**.
* The same filter is applied **across the whole image**.
* Each filter produces a **feature map** (output image showing detected patterns).

### Convolution Operation

* The filter **slides across the image**.

* At each position:

  **output = sum(input × weight)**

* This operation is called **discrete convolution**.

Result:

* CNN extracts **features automatically**.

---

# Padding

Padding defines **how borders of the image are handled**.

Types:

* **Valid padding**

  * No padding added
  * Output image becomes **smaller**

* **Same padding**

  * Adds pixels around the image
  * Output size stays **the same**

* **Full padding**

  * Adds more border pixels
  * Output becomes **larger**

Padding prevents **losing information near edges**.

---

# Stride

**Stride** defines **how far the filter moves each step**.

Example:

* **Stride = 1** → filter moves **1 pixel**
* **Stride = 2** → filter skips pixels → **smaller output**

There is **no fixed rule** for choosing stride; it depends on the task.

---

# Output Size (Simplified)

Output size roughly depends on:

**Input size + Padding − Filter size**

Stride also affects the final dimension.

---

# Pooling Layer

Pooling is used for **downsampling (reducing feature map size)**.

Purpose:

* Reduce **computational cost**
* Reduce **overfitting**
* Keep **important features**

Common types:

### Max Pooling

* Selects the **maximum value** in each region (often **2×2**).
* Keeps the **strongest feature**.

### Average Pooling

* Computes the **average value** in each region.

---

# Flatten Layer

Before the final classifier:

* Feature maps are **flattened into a 1D vector**.
* This vector is passed to **fully connected layers**.

---

# Output Layer

Activation depends on the problem:

* **Sigmoid** → Binary classification
* **Softmax** → Multi-class classification

---

# Loss Functions

* **Binary Cross-Entropy**

  * Used for **two classes**

* **Categorical Cross-Entropy**

  * Used for **multiple classes**

---

# Regularization

To avoid **overfitting**:

### Dropout

* Randomly **deactivates neurons during training**.

### Data Augmentation

Artificially increases dataset size by:

* rotation
* translation
* flipping
* scaling

This helps the model become **invariant to small image changes**.

---

# Typical CNN Pipeline

```
Convolution → ReLU → Pooling
        ↓
Convolution → ReLU → Pooling
        ↓
Flatten
        ↓
Fully Connected Layer
        ↓
Softmax / Sigmoid
```

**ReLU** is commonly used because it introduces **non-linearity**.

---

# Good CNN Design Practices

* First layer may use **larger filters**.
* **Increase number of filters** in deeper layers.
* **Reduce spatial size** as depth increases.
* Use **dropout** to prevent overfitting.
* Use **data augmentation** during preprocessing.

---

# CNN vs Traditional Neural Network

| Model        | Input                                                       |
| ------------ | ----------------------------------------------------------- |
| **CNN**      | Learns **features automatically** (edges, shapes, textures) |
| **Dense NN** | Uses **raw pixel values only**                              |

CNNs are **much better for images**.
