# 01 - Creating random tensors with Tensorflow


```python
random_tensor_1 = tf.random.Generator.from_seed(7)
random_tensor_2 = tf.random.Generator.from_seed(8)
random_tensor_1, random_tensor_2
```

```python
# create random distribution from random tensor

random_tensor_1 = random_tensor_1.normal(shape=(1,2, 3))
random_tensor_2 = random_tensor_2.normal(shape=(1,2, 3))
random_tensor_1, random_tensor_2, random_tensor_1==random_tensor_2
```

# Shuffle tensor in Tensorflow


```python
shuffle = tf.constant([[1,2],[3,4], [5,6]])
shuffle
```

```python
# Shuffle the random



shuffle_random = tf.random.shuffle(shuffle)
shuffle_random
```

# Creating tensor from Numpy

```python
import numpy as np
print(np.__version__)
```

```python
numpy_A = np.ones(shape=(3, 4))
numpy_A
```

```python
# Create numpy array with arange
numpy_B = np.arange(5,12, dtype=np.int32)
numpy_B
```

## Creating tensor from the numpy matrix
> Tensor data is much faster in GPU

```python
A = tf.constant(numpy_A, shape=(2, 3, 2))
A
```

```python
B = tf.constant(numpy_B, dtype=tf.float16)
B
```
